```python
import pandas as pd
import numpy as np
```

```python
train = pd.read_csv(r"Processed Datasets\final_train.csv")

train.head()
```

```python
test = pd.read_csv(r"Processed Datasets\final_test.csv")
test.head()
```

```python
cat_cols = train.drop(columns=['TARGET', 'SK_ID_CURR']).select_dtypes(include=['object', 'str']).columns
cat_cols
```

```python
numeric_cols = train.select_dtypes(include="number")
inf_cols = numeric_cols.columns[np.isinf(numeric_cols).any()]
inf_cols
```

```python
all_nan = train.columns[train.isna().all()]

print(all_nan)
```

```python
train.replace([np.inf, -np.inf], np.nan, inplace=True)
```

```python


summary = []

for col in numeric_cols.columns:

    Q1 = train[col].quantile(0.25)
    Q3 = train[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outlier_mask = (train[col] < lower) | (train[col] > upper)

    outlier_count = outlier_mask.sum()
    outlier_pct = round(outlier_count / len(train) * 100, 2)

    skew = round(train[col].skew(), 2)

    missing_pct = round(train[col].isna().mean() * 100, 2)

    minimum = train[col].min()
    maximum = train[col].max()

    unique = train[col].nunique()

    # ---------- Decision ----------
    if col == "TARGET":
        decision = "Target"
    elif col.startswith("SK_ID"):
        decision = "Ignore (ID)"
    elif train[col].nunique() <= 2:
        decision = "Binary"
    elif col == "DAYS_EMPLOYED":
        decision = "Replace sentinel"
    elif outlier_pct < 5:
        decision = "Keep"
    elif outlier_pct < 20:
        decision = "Review"
    else:
        decision = "Investigate"

    summary.append({
        "Feature": col,
        "Outlier %": outlier_pct,
        "Skew": skew,
        "Missing %": missing_pct,
        "Min": minimum,
        "Max": maximum,
        "Unique": unique,
        "Decision": decision
    })

outlier_summary = (
    pd.DataFrame(summary)
      .sort_values("Outlier %", ascending=False)
      .reset_index(drop=True)
)

outlier_summary
```
